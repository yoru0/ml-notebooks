## Speech To Text

```python
import librosa
import torch
from transformers import Wav2Vec2ForCTC, Wav2Vec2Tokenizer

model = Wav2Vec2ForCTC.from_pretrained("facebook/wav2vec2-base-960h")
tokenizer = Wav2Vec2Tokenizer.from_pretrained("facebook/wav2vec2-base-960h")

audio_path = '../05-wav2vec/speech.wav'

audio, sr = librosa.load(audio_path, sr=16000)

print(audio)

input_values = tokenizer(audio, return_tensors="pt").input_values

print(input_values)

logits = model(input_values).logits

print(logits.shape)

predicted_ids = torch.argmax(logits, dim=-1)

print(predicted_ids)

transcription = tokenizer.batch_decode(predicted_ids)[0]

print(transcription)
```

## Text To Speech

```python
from gtts import gTTS

text = 'THE STALE SMELL OF OLD BEER LINGERS IT TAKES HEAT TO BRING OUT THE ODOR A COLD DIP RESTORES HEALTH AND ZEST A SALT PICKLE TASTES FINE WITH HAM TAKOS AL PASTORE ARE MY FAVORITE A ZESTFUL FOOD IS THE HOT CROSS BUN'

language = 'en'

englishAudio = gTTS(text=text, lang=language, tld='com.au', slow=False)

from pathlib import Path

OUT_DIR = Path.home() / '.cache' / 'ml-notebooks' / 'speech-recognition' / 'text-to-speech'
OUT_DIR.mkdir(parents=True, exist_ok=True)

englishAudio.save(str(OUT_DIR / 'english.mp3'))
```
