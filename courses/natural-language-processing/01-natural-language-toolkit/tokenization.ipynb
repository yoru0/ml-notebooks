```python
# Tokenizing = memecah kalimat menjadi token
# word_tokenize = memecah kalimat atau sekumpulan kata jadi kata
# sent_tokenize = memecah sekumpulan kalimat menjadi sebuah kalimat
```

```python
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
sentence = 'Mr. Smith is leaving for Jakarta at 12 PM today.'
```

```python
word_list = word_tokenize(sentence)
print(word_list)
```

```python
sentences = 'Mr. Smith is leaving for Jakarta at 12 PM today. The train is training. The brain is not braining'
```

```python
sentence_list = sent_tokenize(sentences)
print(sentence_list)
```

```python
# Stopwords = menghilangkan kata-kata yang tidak memiliki makna yang terlalu penting/dibutuhkan
# Ex: I want to eat -> I want eat
```

```python
from nltk.corpus import stopwords
```

```python
eng_stopwords = stopwords.words('english')
print(eng_stopwords)
```

```python
removed_stopwords_list = []

# ['Mr.', 'Smith', 'is', 'leaving', 'for', 'Jakarta', 'at', '12', 'PM', 'today', '.']

for word in word_list:
    if word not in eng_stopwords:
        removed_stopwords_list.append(word)
        
print(removed_stopwords_list)
```

```python
removed_stopwords_list = [word.lower() for word in word_list if word not in eng_stopwords]
print(removed_stopwords_list)
```

```python
import string
```

```python
punctuation_list = string.punctuation
print(punctuation_list)
```

```python
removed_punctuation_words = [word for word in removed_stopwords_list if word not in punctuation_list]
print(removed_punctuation_words)
```

```python
# Remove numbers
# .isdigit
removed_numbers_words = [word for word in removed_punctuation_words if not word.isdigit()]

# removed_numbers_words = [word for word in removed_punctuation_words if word.isalpha()]
# .isalpha tidak akurat karena menghapus gelar yang ada '.'

print(removed_numbers_words)
```
