```python
import nltk
```

```python
from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer, WordNetLemmatizer
```

Stemming: motong kata buat ketemu akarnya

Lemmatization: motong/ganti kata buat ketemu lemmanya (kata asli)

Stemming yang akan digunakan -> Snowball

Lemmatization yang akan digunakan -> WordnetLemmatizer

Lemmatization jauh lebih efektif kalau digabung dengan Part-of-Speech (POS)

```python
# Stemming
words = ['talking', 'playing', 'mice', 'teaching', 'happiness', 'generation', 'generous']

# Porter stemmer
for word in words:
    print(word + ' | ', PorterStemmer().stem(word))
```

```python
# Snowball stemmer
for word in words:
    print(word + ' | ', SnowballStemmer("english").stem(word))
```

```python
# Lancaster stemmer
for word in words:
    print(word + ' | ', LancasterStemmer().stem(word))
```

```python
# Lemmatization

for word in words:
    print(word + ' | ', WordNetLemmatizer().lemmatize(word, 'v'))
    
print(WordNetLemmatizer().lemmatize('talking', 'v'))    # Harus dideskripsikan sebagai verb baru keganti
```

```python
from nltk.corpus import wordnet
```

```python
# Wordnet (Kamus)
print(wordnet.synsets('happy'))

for synonym in wordnet.synsets('dog'):
    print(synonym.name(), "->", synonym.definition(), synonym.examples())
```

```python
# Synonym
synonym = []
for i in wordnet.synsets('happy'):
    for j in i.lemmas():
        synonym.append(j.name())

print('Synonym dari happy:', set(synonym))
```

```python
# Antonym
antonym = []
for i in wordnet.synsets('happy'):
    for j in i.lemmas():
        for k in j.antonyms():
            antonym.append(k.name())

print('Antonym dari happy:', set(antonym))
```

```python
kata = input("Masukkan kata:")
# Buat nampilin definisi, sinonim, antonim

definisi = []
sinonim = []
antonim = []

for i in wordnet.synsets(kata):
    definisi.append(i.definition())
    
    for j in i.lemmas():
        sinonim.append(j.name())
        
        for k in j.antonyms():
            antonim.append(k.name())

# print(set(definisi))
# print(set(sinonim))
# print(set(antonim))

print(f'Kata: {kata}\n')
print(f'Definisi: {set(definisi)}\n')
print(f'Sinonim: {set(sinonim)}\n')
print(f'Antonim: {set(antonim)}\n')
```
