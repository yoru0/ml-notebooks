```python
# POS (Part of Speech) Tag
# Menentukan kategori dari setiap kata
# Noun, adjective, adverb
# NN, JJ
```

```python
from nltk.tag import pos_tag
from nltk.tokenize import word_tokenize
```

```python
sentence = "Mr. Smith is leaving for Jakarta this weekend but Mrs. Smith doesn't know."
```

```python
words =  word_tokenize(sentence)
print(words)
```

```python
tagged = pos_tag(words)
print(tagged)
```

```python
# FreqDist (Frequency Distribution)
# Cari seberapa sering sebuah kata muncul

# I ordered a bowl of chicken rice and a bowl of chicken ramen but I only like the chicken in the ramen.

# most_common -> menampilkan kata-kata yang paling sering muncul
```

```python
from nltk.probability import FreqDist
```

```python
sentence = "I ordered a bowl of chicken rice and a bowl of chicken ramen but I only like the chicken in the ramen."
words = word_tokenize(sentence)

fd = FreqDist(words)
most_common_words = fd.most_common()
print(most_common_words) 
```

```python
for word, count in fd.most_common():
    print(f"Word: {word}, count: {count}")
```

```python
# Print 1 teratas
for word, count in fd.most_common(1):
    print(f"Word: {word}, count: {count}")

print("")

# Print 3 teratas
for word, count in fd.most_common(3):
    print(f"Word: {word}, count: {count}")
```
