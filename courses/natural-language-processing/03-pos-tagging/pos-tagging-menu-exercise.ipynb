```python
#     1. POS Tagging
#     2. Show most common words
#     3. Exit
#     Enter your choice [1-3]:
```

```python
from nltk.tag import pos_tag
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist
```

```python
def POStag():
    sentence = input("Enter your sentence: ")
    words = word_tokenize(sentence)
    
    tagged = pos_tag(words)
    for kata, tag in tagged:
        print(f"Kata: {kata}, Tag: {tag}")
```

```python
def countFreq():
    sentence = input("Enter your sentence: ")
    words = word_tokenize(sentence)
    
    fd = FreqDist(words)
    for kata, count in fd.most_common():
        print(f"Kata: {kata}, Jumlah Kemunculan: {count}")
```

```python
while True:
    print("1. POS Tagging")
    print("2. Show most common words")
    print("3. Exit")
    choice = input("Enter your choice [1-3]: ")
    
    if choice == "1":
        POStag()
        print("")
    elif choice == "2":
        countFreq()
        print("")
    elif choice == "3":
        print("Exiting program...")
        break
    else:
        print("Please input a valid number")
        print("")
```
