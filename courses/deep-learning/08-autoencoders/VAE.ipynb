```python
import shutil
import tempfile
import urllib.request
import zipfile
from pathlib import Path


def download_and_extract(url, dest, rename=None):
    """Download a .zip and extract it into `dest`. Skipped if `dest` already has files.

    `rename(name)` may map each archive member to a new relative path, or return None to skip it.
    """
    dest = Path(dest)
    if dest.exists() and any(dest.iterdir()):
        return dest
    dest.mkdir(parents=True, exist_ok=True)
    with tempfile.NamedTemporaryFile(suffix=".zip", delete=False) as tmp:
        with urllib.request.urlopen(url) as resp:
            shutil.copyfileobj(resp, tmp)
    with zipfile.ZipFile(tmp.name) as zf:
        for member in zf.infolist():
            if member.is_dir():
                continue
            rel = rename(member.filename) if rename else member.filename
            if rel is None:
                continue
            target = dest / rel
            if not target.resolve().is_relative_to(dest.resolve()):
                raise ValueError(f"Unsafe path in archive: {member.filename}")
            target.parent.mkdir(parents=True, exist_ok=True)
            with zf.open(member) as src, open(target, "wb") as out:
                shutil.copyfileobj(src, out)
    Path(tmp.name).unlink()
    return dest

# Shoe vs Sandal vs Boot (15k images) -> ./dataset/{Boot,Sandal,Shoe}
download_and_extract(
    "https://www.kaggle.com/api/v1/datasets/download/hasibalmuzdadid/shoe-vs-sandal-vs-boot-dataset-15k-images",
    "dataset",
    rename=lambda name: Path(*Path(name).parts[1:]),
)
```
