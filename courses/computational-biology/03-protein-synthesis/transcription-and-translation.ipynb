# Transcription and Translation

Convert DNA to RNA (transcription) and RNA to amino acids (translation), inspect the codon table and view a protein structure in 3D.

```python
from Bio.Seq import Seq
```

```python
# DNA sequence = length is a multiple of 3, ATGC
dna_seq = Seq("ATGATCTCGTAA")
print(dna_seq)
```

## From DNA to protein

A gene's DNA is transcribed into mRNA, then translated into a chain of amino acids (a polypeptide). Each group of three nucleotides, a codon, codes for one amino acid. Biopython's `transcribe()` expects the coding strand, so it simply replaces every T with U.

```python
# Transcription -> converts DNA into RNA
# -> replaces Thymine (T) with Uracil (U)

rna_seq = dna_seq.transcribe()
print(rna_seq)
```

```python
# Translation -> converts RNA into amino acids
# -> the nucleotides in the sequence are grouped into codons
amino_acid = rna_seq.translate()
print(amino_acid)
```

## The codon table

The codon table maps each of the 64 codons to an amino acid or a stop signal. `CodonTable` lets you inspect the standard table.

```python
from Bio.Data import CodonTable
```

```python
# Method 1
# codon_table = CodonTable.unambiguous_dna_by_name["Standard"]

# Method 2
codon_table = CodonTable.unambiguous_dna_by_id[1]
print(codon_table)
```

```python
print(dna_seq)
print(amino_acid)
```

## Amino acid names

`seq3` converts one-letter amino acid codes into three-letter names.

```python
from Bio.SeqUtils import seq3
```

```python
polypeptide = seq3(amino_acid)
print(polypeptide)
```

## Viewing a protein in 3D

`py3Dmol` shows a protein structure from the Protein Data Bank (PDB) inside the notebook. It downloads the structure, so it needs internet access.

```python
import py3Dmol
```

```python
# 6LU7 -> molecular model of the SARS-CoV-2 protein -> COVID-19
viewMol = py3Dmol.view(query="pdb:6Lu7")

# .setStyle([dictionary of style options])
viewMol.setStyle({'cartoon': {'color': 'spectrum'}})
viewMol.addStyle({'stick': {'color': 'red'}})
viewMol.show()
```
